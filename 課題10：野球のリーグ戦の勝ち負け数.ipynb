```python
class Baseball_Team:
    name = ""
    win = 0
    lose = 0
    draw = 0

    def __init__(self, name, win, lose, draw):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw

    # method
    def calc_win_rate(self):
        return ( round(self.win / (self.win + self.lose),3))
    
    def show_team_result(self):
        print(f"{self.name:<10} {self.win:>4} {self.lose:>4} {self.draw:>4} {self.calc_win_rate():.3f}" )
```

```python
Archers_Class = Baseball_Team("Archers",77,64,2)
Meteors_Class = Baseball_Team("Meteors",71,69,3)
Cheetahs_Class = Baseball_Team("Cheetahs",69,68,6)
Sharks_Class = Baseball_Team("Sharks",70,70,3)
Pythons_Class = Baseball_Team("Pythons",68,73,2)
Owls_Class = Baseball_Team("Owls",59,82,2)
print(f"{'team':<10} {'win':>4} {'lose':>4} {'draw':>4} {'rate':<5}" )
Archers_Class.show_team_result()
Meteors_Class.show_team_result()
Cheetahs_Class.show_team_result()
Sharks_Class.show_team_result()
Pythons_Class.show_team_result()
Owls_Class.show_team_result()
```
